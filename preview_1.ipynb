## 写一个 Python 程序，分别用 for 和 while 循环实现对一个给定序列的倒排序输出。例如，给定 L=[1,2,3,4,5]，输出为 [5,4,3,2,1]

```python
def reverse_list_for(lst):
    result = []

    for i in range(len(lst)-1, -1, -1):#注意：到0为止，stop为-1
        #倒序遍历，逐个添加至result的末尾（后面先加）
        result.append(lst[i])
        
    return result

user_list = list(map(int, input("请输入以空格分隔的序列：").split()))

print(reverse_list_for(user_list))
```

```python
def reverse_list_while(lst):
    result = []

    i = len(lst)-1
    while i>-1:
        result.append(lst[i])
        i -= 1

    return result

user_list = list(map(int, input("请输入以空格分隔的序列：").split()))

print(reverse_list_while(user_list))
```

## 写一个 Python 程序，判断一个输入的字符串 S 是否包含由两个或两个以上连续出现的相同字符组成的字符串。例如，abccccda 中就包含 cccc 这个由 4 个连续字符组成的字符串

### 使用prev和curr变量迭代比较

```python
def judge_same_chr(S):
    prev = S[0]
    
    for i in range(1, len(S)-1):
        if prev == S[i]:
            print("Yes")
            return
        prev = S[i]

    print("No")

judge_same_chr(input("请输入一个字符串："))#函数没有返回值，无需在print()中调用
```

## 写一个 Python 程序，输入一个字符串 S，去掉其中所有的空格后输出。例如，“DataScience and Engineering”，去掉空格后为“DataScienceandEngineering”。

#### 连接列表中的单个字符串->完整字符串："分隔符"join.(字符串列表)

#### 我的思路：遍历原字符串->去掉空格，存储于新的字符串列表->使用join函数重新连接为单个完整字符串

```python
def delete_space(S):
    result_lst = []

    for i in range(0, len(S), 1):
        if S[i] != " ":
            result_lst.append(S[i])

    return "".join(result_lst)

without_space_S = delete_space(input("请输入一个字符串："))

print(without_space_S)
```

### 更简洁的写法：使用内置replace()替换字符串中特定内容
### 字符串.replace("旧内容", "新内容"， 次数)

```python
def delete_space_S(S):
    spaceless_S = S.replace(" ", "")
    
    return spaceless_S

spaceless = delete_space_S(input("请输入一个字符串："))

print(spaceless)
```

## 13.写一个 Python 程序，给定一个常数 n（n>0），求 n 的阶乘，即 n!=1×2×...×(n-1)×n。例如，4!=24，5!=120

```python
def fact(n):
    if n == 1:
        return 1
    else:
        return n*fact(n-1)

print(fact(int(input("输入一个正常数，求得阶乘："))))
```
