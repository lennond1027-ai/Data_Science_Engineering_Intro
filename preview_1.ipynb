## 编写Python程序，打印"数据科学与工程导论"，并使用 print(chr(0x2605)) 语句将这句话包围起来。

```python
print(chr(0x2605))

print(len(chr(0x2605).encode('utf-8')))
```

### 解答：f-string

```python
star = chr(0x2605)

for i in range(11):
    print(star, end='')


print(f"\n{star}数据科学与工程导论{star}")


for i in range(11):
    print(star, end='')
```

### 标答：占位符

```python
star = chr(0x2605)

for i in range(11):
    print(star, end='')

#format()方法：{}.format(value)
print("\n{}数据科学与工程导论{}".format(star, star))

#老式写法
#print("\n%s数据科学与工程导论%s" %(star, star))


for i in range(11):
    print(star, end='')
```
