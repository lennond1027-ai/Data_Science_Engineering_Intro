## 写一个 Python 程序，输入 x, y, z 这三个数，将这三个数从小到大使用 print 函数打印出来

### 输入一行数字，转换为列表

```python
nums = list(map(int, input("请输入一行数字：").split()))

print("转换后的列表：",nums)

#map() 是 Python 的一个内置函数，用来把一个函数作用到可迭代对象（如列表、字符串、元组等）的每个元素上，返回一个 map 对象（需要用 list() 等转换查看）。
#input()：读取字符串 "1 2 3 4 5"
#.split()：按照指定分隔符切割字符串，并返回一个列表，默认分隔为空格
#map(int, ...)：把字符串转成整数
#list()：转换成列表
```

### 排序
##### 返回新的排序结果，不改变原数据：
#### 可自定义规则：
#### sorted(x, key=规则(eg:len/ lamda x: x[1]))
##### 直接修改原列表，不保留原数据：x.sort()
##### 降序：sorted(x, reverse=True) / x.sort(reverse=Ture)

```python
nums = list(map(int, input("请输入三个整数，以空格分开：").split()))

#print("升序：", nums.sort())
#问题：sort()方法会直接修改原列表，并返回None，而非排序后的列表。
# 所以不能直接打印nums.sort()，而应该先排序，再打印nums。

nums.sort()

print("升序：", nums)

#法二——sorted()：print("升序：", sorted(nums))
```

```python
nums = list(map(int, input("请输入以空格分隔的三个整数：").split()))

print(sorted(nums, reverse=True))
```

## 写一个 Python 程序，有 w, x, y, z 四个数，将这四个数从大到小使用 print 函数打印出来

```python
nums = list(map(int, input("请输入以单空格分隔的四个整数：").split()))

nums.sort(reverse=True)

print("从大到小排序后的结果是: ", nums)
```

### 标答

```python
w = float(input("请输入第一个数: "))
x = float(input("请输入第二个数: "))
y = float(input("请输入第三个数: "))
z = float(input("请输入第四个数: "))

numbers = [w, x, y, z]
numbers.sort(reverse=True)

print("从大到小排序后的结果是: ", numbers)
```

## 写一个 Python 程序，输出 1-100 中的所有奇数

```python
print("1-100之间的奇数：")

for i in range (100):
    if i % 2 == 1:
        print(i, end=' ')
```

## 写一个 Python 程序，用 for 循环，求解 1 到 100 的和

```python
sum = 0

for i in range(100):
    sum += i

print("1-100求和：", sum)
```
