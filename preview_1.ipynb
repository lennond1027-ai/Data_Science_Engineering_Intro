### 解答：f-string
#####  （推荐）：更接近自然语言

```python
star = chr(0x2605)

for i in range(11):
    print(star, end='')


print(f"\n{star}数据科学与工程导论{star}")


for i in range(11):
    print(star, end='')
```

### 标答：占位符

```python
star = chr(0x2605)

for i in range(11):
    print(star, end='')

#format()方法：{}.format(value)
print("\n{}数据科学与工程导论{}".format(star, star))

#老式写法
#print("\n%s数据科学与工程导论%s" %(star, star))


for i in range(11):
    print(star, end='')
```

### 输入一行数字，转换为列表

```python
nums = list(map(int, input("请输入一行数字：").split()))

print("转换后的列表：",nums)

#map() 是 Python 的一个内置函数，用来把一个函数作用到可迭代对象（如列表、字符串、元组等）的每个元素上，返回一个 map 对象（需要用 list() 等转换查看）。
#input()：读取字符串 "1 2 3 4 5"
#.split()：按照指定分隔符切割字符串，并返回一个列表，默认分隔为空格
#map(int, ...)：把字符串转成整数
#list()：转换成列表
```
